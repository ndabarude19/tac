# Extraction de Keywords

## Imports

```python
import os
import yake
from collections import Counter
```

## Extraire les mots clés des articles du Soir de 1960 avec Yake

https://github.com/INESCTEC/yake

```python
# Instantier l'extracteur de mots clés
kw_extractor = yake.KeywordExtractor(lan="fr", top=50)
kw_extractor
```

```python
# Choisir une année
year = 1960
# Lister les fichiers de cette année
data_path = "../../data/txt/"
files = sorted([f for f in os.listdir(data_path) if f.endswith('.txt') and f"_{year}-" in f])
```

```python
# Imprimer le nombre de fichiers identifiés
len(files)
```

```python
# Les dix premiers fichiers
files[:10]
```

```python
# Choisir un fichier
this_file = files[0]
this_file
```

```python
# Récupérer le texte du fichier
with open(os.path.join(data_path, this_file), 'r', encoding='utf-8') as fh:
    text = fh.read()
text[:500]
```

```python
# Extraire les mots clés de ce texte
keywords = kw_extractor.extract_keywords(text)
keywords
```

```python
# Ne garder que les bigrammes
kept = []
for kw, score in keywords:
    words = kw.split()
    if len(words) == 2:
        kept.append(kw)
kept
```

## Faire la même opération sur tous les documents

```python
for f in files:
    with open(os.path.join(data_path, f), 'r', encoding="utf-8") as fh:
        text = fh.read()
    keywords = kw_extractor.extract_keywords(text)
    kept = []
    for kw, score in keywords:
        words = kw.split()
        if len(words) == 2:
            kept.append(kw)
    print(f"{f} mentions these keywords: {', '.join(kept)}...")
```

## Synthèse : les bigrammes les plus fréquents en 1960

```python
# Compter dans combien de numéros chaque bigramme apparaît parmi les mots-clés
year_keywords = Counter()
for f in files:
    with open(os.path.join(data_path, f), 'r', encoding='utf-8') as fh:
        text = fh.read()
    for kw, score in kw_extractor.extract_keywords(text):
        if len(kw.split()) == 2:
            year_keywords[kw] += 1

year_keywords.most_common(30)
```
