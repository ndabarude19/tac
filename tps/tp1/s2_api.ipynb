# Enrichissement de CAMille avec l'API Wikipédia « Ce jour-là »
à partir de la date d'un journal, ce script retrouve les événements historiques de ce jour sur Wikipédia.

## Imports

```python

import requests
```

```python
fichier = "KB_JB230_1892-08-07_01-0003.pdf"
date = fichier.split("_")[2]
annee, mois, jour = date.split("-")
print(annee, mois, jour)
```

## Utiliser l'API Wikipédia « Ce jour-là » pour retrouver les événements à la date d'un journal

```python
headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_10_1) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/39.0.2171.95 Safari/537.36'}
url = f"https://fr.wikipedia.org/api/rest_v1/feed/onthisday/events/{mois}/{jour}"
resp = requests.get(url, headers=headers)
evenements = resp.json()["events"]
print(len(evenements), "événements trouvés")
```

## Afficher les événements historiques du jour

```python
for evenement in evenements:
    print(evenement["year"], ":", evenement["text"])
```

```python
trouve = False
for evenement in evenements:
    if str(evenement["year"]) == annee:
        print(evenement["year"], ":", evenement["text"])
        trouve = True

if not trouve:
    print(f"Aucun événement recensé par Wikipédia le {jour}/{mois}/{annee}.")
```
